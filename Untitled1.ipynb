```python
# Declaración inicial
precio = 100

precio -= 20    # Aplica descuento: precio = 100 - 20 (80)
precio *= 1.16  # Agrega 16% IVA:  precio = 80 * 1.16 (92.8)
precio /= 2     # Divide entre 2:  precio = 92.8 / 2 (46.4)

print(f"Precio final: ${precio}")
```

```python
 # EJERCICIOS WHILE
 # 1 Calcular la suma de los números impares del 1 al 50.
 # 2 Imprimir los múltiplos de 3 menores que 100 utilizando un bucle while
 #3 Imprimir los números primos menores que 100 utilizando un bucle while
```

```python
# Calcular la suma de los números impares del 1 al 50
suma = 0
numero = 1
while numero <= 50:
    if numero % 2 != 0:
        suma += numero
    numero += 1
print(f"La suma de los números impares del 1 al 50 es: {suma}")
```

```python
# Imprimir los múltiplos de 3 menores que 100 utilizando un bucle while
numero = 3
while numero < 100:
    print(numero)
    numero += 3
```

```python
# Imprimir los números primos menores que 100 utilizando un bucle while
numero = 2
while numero < 100:
    es_primo = True
    divisor = 2
    while divisor * divisor <= numero:
        if numero % divisor == 0:
            es_primo = False
            break
        divisor += 1
    if es_primo:
        print(numero)
    numero += 1
```

```python
# 5 Calcular el factorial de un número dado
numero = int(input("Ingrese un número: "))
factorial = 1
for i in range(1, numero + 1):
    factorial *= i

print(f"El factorial de {numero} es: {factorial}")
```
